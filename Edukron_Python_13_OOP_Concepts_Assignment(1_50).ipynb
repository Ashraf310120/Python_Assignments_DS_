## === Q1. Create a account class for banking case 1; use sample reference ACC-1001 and explain the result. ===

```python
class Account:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.__balance = balance

    def deposit(self, amount):
        self.__balance = self.__balance + amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance = self.__balance - amount
        else:
            print("Insufficient balance")

    def show_balance(self):
        print("Current Balance:", self.__balance)


account = Account("ACC-1001", 10000)

print("Account Number:", account.account_number)
account.show_balance()

account.deposit(2000)
account.show_balance()

account.withdraw(3000)
account.show_balance()
```

## === Q41. Encapsulate balance reservation for banking case 1; use sample reference ACC-1041 and explain the result. ===

```python
class Account:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.__balance = balance
        self.__reserved = 0

    def reserve_balance(self, amount):
        if amount <= 0:
            print("Invalid reservation amount")
        elif amount > self.__balance - self.__reserved:
            print("Insufficient available balance")
        else:
            self.__reserved = self.__reserved + amount
            print("Amount reserved:", amount)

    def release_balance(self, amount):
        if amount <= self.__reserved:
            self.__reserved = self.__reserved - amount
            print("Reservation released:", amount)
        else:
            print("Invalid release amount")

    def show_balance(self):
        print("Account Number:", self.account_number)
        print("Balance:", self.__balance)
        print("Reserved Amount:", self.__reserved)
        print("Available Balance:", self.__balance - self.__reserved)


account = Account("ACC-1041", 10000)

account.reserve_balance(3000)

account.show_balance()

account.release_balance(1000)

account.show_balance()
```

## === Q42. Model aggregation between branch and employees for banking case 1; use sample reference ACC-1042 and explain the result. ===

```python
class Employee:
    def __init__(self, employee_id, name):
        self.employee_id = employee_id
        self.name = name

    def show_employee(self):
        print("Employee ID:", self.employee_id)
        print("Employee Name:", self.name)


class Branch:
    def __init__(self, branch_name):
        self.branch_name = branch_name
        self.employees = []

    def add_employee(self, employee):
        if not isinstance(employee, Employee):
            print("Invalid employee")
        else:
            self.employees.append(employee)
            print("Employee added")

    def show_branch(self):
        print("Branch:", self.branch_name)
        print("Employees:")

        if len(self.employees) == 0:
            print("No employees")
        else:
            for employee in self.employees:
                employee.show_employee()


employee1 = Employee("EMP-1042", "Rahul")
employee2 = Employee("EMP-1043", "Aman")

branch = Branch("Main Branch")

branch.show_branch()

branch.add_employee(employee1)
branch.add_employee(employee2)

branch.show_branch()

branch.add_employee("Invalid Employee")
```

```python
class Account:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            return True
        return False

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            return True
        return False

    def show_balance(self):
        return self.__balance


class AccountRepository:
    def __init__(self):
        self.accounts = {}

    def save(self, account):
        if isinstance(account, Account):
            self.accounts[account.account_number] = account
            print("Account saved")
        else:
            print("Invalid account")

    def find(self, account_number):
        return self.accounts.get(account_number)


repository = AccountRepository()

account = Account("ACC-1044", 10000)

repository.save(account)

found_account = repository.find("ACC-1044")

if found_account:
    print("Account Number:", found_account.account_number)
    print("Balance:", found_account.show_balance())
else:
    print("Account not found")

missing_account = repository.find("ACC-9999")

if missing_account:
    print("Account found")
else:
    print("Account not found")
```

## === Q48. Use polymorphic payment processors for banking case 2; use sample reference ACC-1048 and explain the result. ===

## === Q49. Compose transfer from transferline objects for banking case 2; use sample reference ACC-1049 and explain the result. ===

```python
class TransferLine:
    def __init__(self, account_number, amount):
        self.account_number = account_number
        self.amount = amount

    def show_line(self):
        print("Account:", self.account_number)
        print("Amount:", self.amount)


class Transfer:
    def __init__(self, transfer_id):
        self.transfer_id = transfer_id
        self.lines = []

    def add_line(self, line):
        if isinstance(line, TransferLine):
            self.lines.append(line)
            print("Transfer line added")
        else:
            print("Invalid transfer line")

    def show_transfer(self):
        print("Transfer ID:", self.transfer_id)

        if len(self.lines) == 0:
            print("No transfer lines")
        else:
            for line in self.lines:
                line.show_line()


transfer = Transfer("ACC-1049")

line1 = TransferLine("ACC-1001", 5000)
line2 = TransferLine("ACC-1002", 5000)

transfer.show_transfer()

transfer.add_line(line1)
transfer.add_line(line2)

transfer.show_transfer()

transfer.add_line("Invalid Line")
```

## === Q50. Define an abstract paymentmethod for banking case 2; use sample reference ACC-1050 and explain the result. ===

```python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class CardPayment(PaymentMethod):
    def pay(self, amount):
        if amount > 0:
            print("Card payment successful:", amount)
        else:
            print("Invalid payment amount")


class UpiPayment(PaymentMethod):
    def pay(self, amount):
        if amount > 0:
            print("UPI payment successful:", amount)
        else:
            print("Invalid payment amount")


card = CardPayment()
upi = UpiPayment()

card.pay(500)
upi.pay(1000)

card.pay(0)
upi.pay(-200)
```
