## === Q33. Group accounts using a helper function for banking case 1; use sample reference ACC-1033 and explain the result. ===

```python
def get_group(account):
    return account["account_type"]


def group_accounts(accounts):
    groups = {}

    for account in accounts:
        group = get_group(account)

        if group not in groups:
            groups[group] = []

        groups[group].append(account)

    return groups


# Sample reference: ACC-1033
accounts = [
    {"account_id": "ACC-1033", "account_type": "Savings", "balance": 15000},
    {"account_id": "ACC-1034", "account_type": "Current", "balance": 10000},
    {"account_id": "ACC-1035", "account_type": "Savings", "balance": 20000}
]

result = group_accounts(accounts)

print("Account:", "ACC-1033")
print("Grouped Accounts:", result)
```

## === Q35. Write a decorator that logs checkout calls for banking case 1; use sample reference ACC-1035 and explain the result. ===

```python
def log_checkout(func):
    def wrapper(account_id, amount):
        print("Checkout called for:", account_id)
        result = func(account_id, amount)
        print("Checkout completed")
        return result

    return wrapper


@log_checkout
def checkout(account_id, amount):
    return amount


# Sample reference: ACC-1035
result = checkout("ACC-1035", 5000)

print("Checkout Amount:", result)
```

## === Q36. Build a map-filter-reduce transactions pipeline for banking case 1; use sample reference ACC-1036 and explain the result. ===

```python
from functools import reduce


def process_transactions(transactions):
    # Filter successful transactions
    filtered = filter(lambda t: t["status"] == "success", transactions)

    # Map to transaction amounts
    amounts = map(lambda t: t["amount"], filtered)

    # Reduce to get the total
    total = reduce(lambda x, y: x + y, amounts, 0)

    return total


# Sample reference: ACC-1036
transactions = [
    {"account_id": "ACC-1036", "amount": 1000, "status": "success"},
    {"account_id": "ACC-1036", "amount": 500, "status": "failed"},
    {"account_id": "ACC-1036", "amount": 1500, "status": "success"}
]

result = process_transactions(transactions)

print("Account:", "ACC-1036")
print("Total Successful Transactions:", result)
```

## === Q40. Implement recursive category total for banking case 1; use sample reference ACC-1040 and explain the result. ===

```python
def category_total(categories):
    if not categories:
        return 0

    first = categories[0]
    remaining = categories[1:]

    return first["amount"] + category_total(remaining)


# Sample reference: ACC-1040
categories = [
    {"account_id": "ACC-1040", "category": "Deposit", "amount": 2000},
    {"account_id": "ACC-1040", "category": "Loan", "amount": 3000},
    {"account_id": "ACC-1040", "category": "Interest", "amount": 1000}
]

result = category_total(categories)

print("Account:", "ACC-1040")
print("Category Total:", result)
```

## === Q41. Validate arguments with a wrapper for banking case 1; use sample reference ACC-1041 and explain the result. ===

```python
def validate_arguments(func):
    def wrapper(account_id, amount):
        if not isinstance(account_id, str):
            raise TypeError("Account ID must be a string")

        if not isinstance(amount, (int, float)):
            raise TypeError("Amount must be a number")

        if amount < 0:
            raise ValueError("Amount cannot be negative")

        return func(account_id, amount)

    return wrapper


@validate_arguments
def deposit(account_id, amount):
    return f"Deposit of ₹{amount} accepted for {account_id}"


# Sample reference: ACC-1041
result = deposit("ACC-1041", 5000)

print(result)
```

## === Q44. Calculate basket metrics in one pass for banking case 1; use sample reference ACC-1044 and explain the result. ===

```python
def basket_metrics(amounts):
    total = 0
    count = 0

    for amount in amounts:
        total += amount
        count += 1

    if count == 0:
        return 0, 0, 0

    average = total / count
    return total, count, average


# Sample reference: ACC-1044
basket = [500, 1000, 1500, 2000]

total, count, average = basket_metrics(basket)

print("Account:", "ACC-1044")
print("Total:", total)
print("Count:", count)
print("Average:", average)
```
