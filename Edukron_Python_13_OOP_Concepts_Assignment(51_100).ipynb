## === Q56. Encapsulate balance reservation for banking case 2; use sample reference ACC-1056 and explain the result. ===

```python
class Account:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.__balance = balance
        self.__reserved = 0

    def reserve_balance(self, amount):
        if not isinstance(amount, (int, float)):
            print("Invalid amount")
        elif amount <= 0:
            print("Invalid reservation amount")
        elif amount > self.__balance - self.__reserved:
            print("Insufficient available balance")
        else:
            self.__reserved += amount
            print("Amount reserved:", amount)

    def release_balance(self, amount):
        if not isinstance(amount, (int, float)):
            print("Invalid amount")
        elif amount <= 0 or amount > self.__reserved:
            print("Invalid release amount")
        else:
            self.__reserved -= amount
            print("Reservation released:", amount)

    def show_balance(self):
        print("Account Number:", self.account_number)
        print("Balance:", self.__balance)
        print("Reserved Amount:", self.__reserved)
        print("Available Balance:", self.__balance - self.__reserved)


account = Account("ACC-1056", 10000)

account.reserve_balance(3000)
account.show_balance()

account.reserve_balance(8000)

account.release_balance(1000)
account.show_balance()

account.release_balance(5000)

account.reserve_balance(0)
account.reserve_balance("500")
```

## === Q59. Implement a repository abstraction for banking case 2; use sample reference ACC-1059 and explain the result. ===

```python
class Account:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            return True
        return False

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            return True
        return False

    def show_balance(self):
        return self.__balance


class AccountRepository:
    def __init__(self):
        self.accounts = {}

    def save(self, account):
        if isinstance(account, Account):
            self.accounts[account.account_number] = account
            print("Account saved")
        else:
            print("Invalid account")

    def find(self, account_number):
        return self.accounts.get(account_number)


repository = AccountRepository()

account = Account("ACC-1059", 10000)

repository.save(account)

found_account = repository.find("ACC-1059")

if found_account:
    print("Account Number:", found_account.account_number)
    print("Balance:", found_account.show_balance())
else:
    print("Account not found")

missing_account = repository.find("ACC-9999")

if missing_account:
    print("Account found")
else:
    print("Account not found")

repository.save("Invalid Account")
```

## === Q71. Design an extensible pricing engine for banking case 1; use sample reference ACC-1071 and explain the result. ===

```python
class PricingRule:
    def calculate_price(self, amount):
        return amount


class RegularPricing(PricingRule):
    def calculate_price(self, amount):
        return amount


class PremiumPricing(PricingRule):
    def calculate_price(self, amount):
        return amount * 0.95


class PricingEngine:
    def __init__(self, pricing_rule):
        self.pricing_rule = pricing_rule

    def get_price(self, amount):
        if not isinstance(amount, (int, float)):
            print("Invalid amount")
            return None

        if amount < 0:
            print("Amount cannot be negative")
            return None

        return self.pricing_rule.calculate_price(amount)


regular_rule = RegularPricing()
premium_rule = PremiumPricing()

regular_engine = PricingEngine(regular_rule)
premium_engine = PricingEngine(premium_rule)

print("Account:", "ACC-1071")
print("Regular Price:", regular_engine.get_price(1000))
print("Premium Price:", premium_engine.get_price(1000))

print("Invalid Price:", premium_engine.get_price(-100))
print("Invalid Type:", premium_engine.get_price("1000"))
```

## === Q72. Implement an observer for low balance for banking case 1; use sample reference ACC-1072 and explain the result. ===

```python
class LowBalanceObserver:
    def update(self, account_number, balance):
        print("Low balance alert for:", account_number)
        print("Current Balance:", balance)


class Account:
    def __init__(self, account_number, balance, limit):
        self.account_number = account_number
        self.__balance = balance
        self.limit = limit
        self.observers = []

    def add_observer(self, observer):
        if isinstance(observer, LowBalanceObserver):
            self.observers.append(observer)
        else:
            print("Invalid observer")

    def withdraw(self, amount):
        if not isinstance(amount, (int, float)):
            print("Invalid amount")
            return

        if amount <= 0:
            print("Invalid withdrawal amount")
            return

        if amount > self.__balance:
            print("Insufficient balance")
            return

        self.__balance -= amount
        print("Withdrawal successful:", amount)

        if self.__balance <= self.limit:
            for observer in self.observers:
                observer.update(self.account_number, self.__balance)

    def show_balance(self):
        print("Account Number:", self.account_number)
        print("Balance:", self.__balance)


account = Account("ACC-1072", 5000, 2000)

observer = LowBalanceObserver()

account.add_observer(observer)

account.show_balance()

account.withdraw(2500)

account.withdraw(100)

account.withdraw(5000)

account.withdraw(-100)
```

## === Q73. Build a state machine for transfer status for banking case 1; use sample reference ACC-1073 and explain the result. ===

```python
class Transfer:
    def __init__(self, transfer_id):
        self.transfer_id = transfer_id
        self.status = "Pending"

    def approve(self):
        if self.status == "Pending":
            self.status = "Approved"
            print("Transfer approved")
        else:
            print("Cannot approve transfer")

    def process(self):
        if self.status == "Approved":
            self.status = "Processing"
            print("Transfer processing")
        else:
            print("Transfer must be approved first")

    def complete(self):
        if self.status == "Processing":
            self.status = "Completed"
            print("Transfer completed")
        else:
            print("Transfer must be processing first")

    def cancel(self):
        if self.status in ["Pending", "Approved"]:
            self.status = "Cancelled"
            print("Transfer cancelled")
        else:
            print("Cannot cancel transfer")

    def show_status(self):
        print("Transfer ID:", self.transfer_id)
        print("Status:", self.status)


transfer = Transfer("ACC-1073")

transfer.show_status()

transfer.process()

transfer.approve()
transfer.process()
transfer.complete()

transfer.show_status()

transfer.cancel()
```

## === Q74. Create an immutable money value object for banking case 1; use sample reference ACC-1074 and explain the result. ===

```python
from dataclasses import dataclass


@dataclass(frozen=True)
class Money:
    amount: float
    currency: str

    def add(self, other):
        if self.currency != other.currency:
            raise ValueError("Currency mismatch")

        return Money(
            self.amount + other.amount,
            self.currency
        )

    def subtract(self, other):
        if self.currency != other.currency:
            raise ValueError("Currency mismatch")

        return Money(
            self.amount - other.amount,
            self.currency
        )


money1 = Money(1000, "INR")
money2 = Money(500, "INR")

total = money1.add(money2)
remaining = money1.subtract(money2)

print("Account:", "ACC-1074")
print("Money 1:", money1)
print("Money 2:", money2)
print("Total:", total)
print("Remaining:", remaining)

try:
    money1.amount = 2000
except Exception as e:
    print("Cannot modify Money:", e)

try:
    dollar = Money(100, "USD")
    money1.add(dollar)
except ValueError as e:
    print("Error:", e)
```

## === Q75. Implement a unit-of-work transaction for banking case 1; use sample reference ACC-1075 and explain the result. ===

```python
class LowBalanceObserver:
    def update(self, account_number, balance):
        print("Low balance alert for:", account_number)
        print("Current Balance:", balance)


class Account:
    def __init__(self, account_number, balance, limit):
        self.account_number = account_number
        self.__balance = balance
        self.limit = limit
        self.observers = []

    def add_observer(self, observer):
        if isinstance(observer, LowBalanceObserver):
            self.observers.append(observer)
        else:
            print("Invalid observer")

    def withdraw(self, amount):
        if not isinstance(amount, (int, float)):
            print("Invalid amount")
            return False

        if amount <= 0:
            print("Invalid withdrawal amount")
            return False

        if amount > self.__balance:
            print("Insufficient balance")
            return False

        self.__balance -= amount
        print("Withdrawal successful:", amount)

        if self.__balance <= self.limit:
            for observer in self.observers:
                observer.update(self.account_number, self.__balance)

        return True

    def deposit(self, amount):
        if not isinstance(amount, (int, float)) or amount <= 0:
            print("Invalid deposit amount")
            return False

        self.__balance += amount
        print("Deposit successful:", amount)
        return True

    def show_balance(self):
        print("Account Number:", self.account_number)
        print("Balance:", self.__balance)


account = Account("ACC-1092", 5000, 2000)

observer = LowBalanceObserver()

account.add_observer(observer)

account.show_balance()

account.withdraw(2500)

account.withdraw(100)

account.deposit(1000)

account.withdraw(5000)

account.withdraw(-100)

account.withdraw("500")

account.add_observer("Invalid Observer")
```

## === Q93. Build a state machine for transfer status for banking case 3; use sample reference ACC-1093 and explain the result. ===

```python
class Transfer:
    def __init__(self, transfer_id):
        self.transfer_id = transfer_id
        self.status = "Pending"

    def approve(self):
        if self.status == "Pending":
            self.status = "Approved"
            print("Transfer approved")
        else:
            print("Cannot approve transfer")

    def process(self):
        if self.status == "Approved":
            self.status = "Processing"
            print("Transfer processing")
        else:
            print("Transfer must be approved first")

    def complete(self):
        if self.status == "Processing":
            self.status = "Completed"
            print("Transfer completed")
        else:
            print("Transfer must be processing first")

    def cancel(self):
        if self.status in ["Pending", "Approved"]:
            self.status = "Cancelled"
            print("Transfer cancelled")
        else:
            print("Cannot cancel transfer")

    def show_status(self):
        print("Transfer ID:", self.transfer_id)
        print("Status:", self.status)


transfer = Transfer("ACC-1093")

transfer.show_status()

transfer.process()

transfer.approve()
transfer.process()
transfer.complete()

transfer.show_status()

transfer.cancel()

invalid_transfer = Transfer("")

invalid_transfer.process()
invalid_transfer.complete()
```

## === Q94. Create an immutable money value object for banking case 3; use sample reference ACC-1094 and explain the result. ===

```python
from dataclasses import dataclass


@dataclass(frozen=True)
class Money:
    amount: float
    currency: str

    def add(self, other):
        if self.currency != other.currency:
            raise ValueError("Currency mismatch")

        return Money(
            self.amount + other.amount,
            self.currency
        )

    def subtract(self, other):
        if self.currency != other.currency:
            raise ValueError("Currency mismatch")

        return Money(
            self.amount - other.amount,
            self.currency
        )


money1 = Money(1000, "INR")
money2 = Money(500, "INR")

total = money1.add(money2)
remaining = money1.subtract(money2)

print("Account:", "ACC-1094")
print("Money 1:", money1)
print("Money 2:", money2)
print("Total:", total)
print("Remaining:", remaining)

try:
    money1.amount = 2000
except Exception as e:
    print("Cannot modify Money:", e)

try:
    dollar = Money(100, "USD")
    money1.add(dollar)
except ValueError as e:
    print("Error:", e)

try:
    empty_money = Money(0, "INR")
    print("Empty Money:", empty_money)
except Exception as e:
    print("Error:", e)
```

## === Q95. Implement a unit-of-work transaction for banking case 3; use sample reference ACC-1095 and explain the result. ===

```python
class Account:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.__balance = balance

    def withdraw(self, amount):
        if not isinstance(amount, (int, float)):
            return False

        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            return True

        return False

    def deposit(self, amount):
        if not isinstance(amount, (int, float)):
            return False

        if amount > 0:
            self.__balance += amount
            return True

        return False

    def get_balance(self):
        return self.__balance


class UnitOfWork:
    def __init__(self):
        self.operations = []

    def add_operation(self, operation):
        if callable(operation):
            self.operations.append(operation)
        else:
            print("Invalid operation")

    def commit(self):
        if len(self.operations) == 0:
            print("No operations")
            return False

        for operation in self.operations:
            if not operation():
                print("Transaction failed")
                return False

        print("Transaction completed")
        return True


from_account = Account("ACC-1095", 10000)
to_account = Account("ACC-1002", 5000)

amount = 3000

transaction = UnitOfWork()

transaction.add_operation(
    lambda: from_account.withdraw(amount)
)

transaction.add_operation(
    lambda: to_account.deposit(amount)
)

transaction.commit()

print("From Account Balance:", from_account.get_balance())
print("To Account Balance:", to_account.get_balance())


failed_transaction = UnitOfWork()

failed_transaction.add_operation(
    lambda: from_account.withdraw(20000)
)

failed_transaction.add_operation(
    lambda: to_account.deposit(20000)
)

failed_transaction.commit()

print("From Account Balance:", from_account.get_balance())
print("To Account Balance:", to_account.get_balance())


empty_transaction = UnitOfWork()
empty_transaction.commit()

transaction.add_operation("Invalid Operation")
```

## === Q97. Design a plugin-ready payment architecture for banking case 3; use sample reference ACC-1097 and explain the result. ===

```python
class PaymentPlugin:
    def pay(self, amount):
        raise NotImplementedError("Payment method must implement pay()")


class CardPayment(PaymentPlugin):
    def pay(self, amount):
        if amount > 0:
            print("Card payment successful:", amount)
            return True
        print("Invalid card payment amount")
        return False


class UpiPayment(PaymentPlugin):
    def pay(self, amount):
        if amount > 0:
            print("UPI payment successful:", amount)
            return True
        print("Invalid UPI payment amount")
        return False


class WalletPayment(PaymentPlugin):
    def pay(self, amount):
        if amount > 0:
            print("Wallet payment successful:", amount)
            return True
        print("Invalid wallet payment amount")
        return False


class PaymentManager:
    def __init__(self):
        self.plugins = {}

    def add_plugin(self, name, plugin):
        if isinstance(plugin, PaymentPlugin):
            self.plugins[name] = plugin
            print("Plugin added:", name)
        else:
            print("Invalid payment plugin")

    def make_payment(self, name, amount):
        if name not in self.plugins:
            print("Payment plugin not found")
            return False

        if not isinstance(amount, (int, float)):
            print("Invalid amount")
            return False

        return self.plugins[name].pay(amount)


manager = PaymentManager()

manager.add_plugin("card", CardPayment())
manager.add_plugin("upi", UpiPayment())
manager.add_plugin("wallet", WalletPayment())

manager.make_payment("card", 500)
manager.make_payment("upi", 1000)
manager.make_payment("wallet", 750)

manager.make_payment("card", -100)
manager.make_payment("cash", 500)
manager.make_payment("upi", "500")

manager.add_plugin("cash", "Invalid Plugin")
```

## === Q98. Implement a descriptor for positive amounts for banking case 3; use sample reference ACC-1098 and explain the result. ===

```python
class PositiveAmount:
    def __set_name__(self, owner, name):
        self.name = "_" + name

    def __get__(self, instance, owner):
        if instance is None:
            return self

        return getattr(instance, self.name)

    def __set__(self, instance, value):
        if not isinstance(value, (int, float)):
            raise TypeError("Amount must be a number")

        if value <= 0:
            raise ValueError("Amount must be positive")

        setattr(instance, self.name, value)


class Transaction:
    amount = PositiveAmount()

    def __init__(self, transaction_id, amount):
        self.transaction_id = transaction_id
        self.amount = amount

    def show_transaction(self):
        print("Transaction ID:", self.transaction_id)
        print("Amount:", self.amount)


transaction = Transaction("ACC-1098", 5000)

transaction.show_transaction()

transaction.amount = 1000
transaction.show_transaction()

try:
    transaction.amount = 0
except ValueError as e:
    print("Error:", e)

try:
    transaction.amount = -500
except ValueError as e:
    print("Error:", e)

try:
    transaction.amount = "500"
except TypeError as e:
    print("Error:", e)
```
