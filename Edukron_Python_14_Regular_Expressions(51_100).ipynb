## === Q51. Validate fee waiver start and end anchors for banking case 2; use sample reference ACC-1051 and explain the result. ===

```python
import re

# Sample reference: ACC-1051

pattern = re.compile(
    r"^(?P<start>START)\s*-\s*(?P<end>END)$"
)

valid_samples = [
    "START-END",
    "START - END",
    "START    -    END"
]

invalid_samples = [
    "",
    None,
    "START",
    "END-START",
    "start-END",
    "START-END-extra"
]

for text in valid_samples:
    match = pattern.fullmatch(text)

    if match:
        print("Valid fee waiver:", match.group("start"), match.group("end"))
    else:
        print("Invalid fee waiver:", text)

for text in invalid_samples:
    if not isinstance(text, str):
        print("Invalid input:", text)
        continue

    match = pattern.fullmatch(text)

    if match:
        print("Valid fee waiver:", match.group("start"), match.group("end"))
    else:
        print("Invalid fee waiver:", text)
```

## === Q52. Redact emails from support text for banking case 2; use sample reference ACC-1052 and explain the result. ===

```python
import re

# Sample reference: ACC-1052

pattern = re.compile(
    r"(?P<email>[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,})"
)

valid_samples = [
    "Customer email is customer@example.com",
    "Please contact support@bank.com for help",
    "Send the receipt to user123@banking.in"
]

invalid_samples = [
    "",
    None,
    "No email address provided",
    "customer@"
]


def redact_emails(text):
    if not isinstance(text, str):
        return None

    return pattern.sub("[REDACTED]", text)


for text in valid_samples:
    result = redact_emails(text)

    if result is not None:
        print("Original:", text)
        print("Redacted:", result)
    else:
        print("Invalid input:", text)

for text in invalid_samples:
    result = redact_emails(text)

    if result is None:
        print("Invalid input:", text)
    else:
        print("Redacted:", result)
```

## === Q58. Replace dates with iso format for banking case 2; use sample reference ACC-1058 and explain the result. ===

```python
import re

# Sample reference: ACC-1058

pattern = re.compile(
    r"(?P<day>\d{2})[-/](?P<month>\d{2})[-/](?P<year>\d{4})"
)

valid_samples = [
    "Receipt date: 15/09/2026",
    "Transaction date: 01-01-2025",
    "Payment made on 31/12/2024"
]

invalid_samples = [
    "",
    None,
    "No date available",
    "15/9/2026",
    "2026-09-15"
]


def replace_date(text):
    if not isinstance(text, str):
        return None

    return pattern.sub(
        lambda match: f"{match.group('year')}-{match.group('month')}-{match.group('day')}",
        text
    )


for text in valid_samples:
    result = replace_date(text)

    if result is not None:
        print("Original:", text)
        print("ISO format:", result)
    else:
        print("Invalid input:", text)

for text in invalid_samples:
    result = replace_date(text)

    if result is None:
        print("Invalid input:", text)
    else:
        print("Result:", result)
```

### === Q64. Normalize account holder phone numbers for banking case 3; use sample reference ACC-1064 and explain the result. ===

```python
import re

# Sample reference: ACC-1064

pattern = re.compile(r"(?P<phone>[6-9]\d{9})")

valid_samples = [
    "9876543210",
    "98765 43210",
    "98765-43210",
    "+91 9876543210"
]

invalid_samples = [
    "",
    None,
    "1234567890",
    "987654321",
    "98765432101",
    "98765ABCDE"
]


def normalize_phone(phone):
    if not isinstance(phone, str):
        return None

    phone = re.sub(r"[ -]", "", phone)

    if phone.startswith("+91"):
        phone = phone[3:]

    match = pattern.fullmatch(phone)

    if match:
        return match.group("phone")

    return None


for phone in valid_samples:
    result = normalize_phone(phone)

    if result:
        print("Normalized phone:", result)
    else:
        print("Invalid phone number:", phone)

for phone in invalid_samples:
    result = normalize_phone(phone)

    if result:
        print("Normalized phone:", result)
    else:
        print("Invalid phone number:", phone)
```
