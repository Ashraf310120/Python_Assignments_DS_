## === Q1. Find a account code for banking case 1; use sample reference ACC-1001 and explain the result. ===

```python
import re

pattern = re.compile(r"(?P<account_code>ACC-\d{4})")

valid_accounts = [
    "ACC-1001",
    "ACC-1234"
]

invalid_accounts = [
    "ACC-123",
    "ACC-12345",
    "acc-1001",
    "",
    None
]

for account in valid_accounts:
    match = pattern.fullmatch(account)

    if match:
        print("Valid account code:", match.group("account_code"))
    else:
        print("Invalid account code:", account)

for account in invalid_accounts:
    if not isinstance(account, str):
        print("Invalid input:", account)
        continue

    match = pattern.fullmatch(account)

    if match:
        print("Valid account code:", match.group("account_code"))
    else:
        print("Invalid account code:", account)
```

## === Q3. Extract all amounts for banking case 1; use sample reference ACC-1003 and explain the result. ===

## === Q7. Split a comma-separated tag list for banking case 1; use sample reference ACC-1007 and explain the result. ===

```python
import re

# Sample reference: ACC-1007

pattern = re.compile(r"\s*,\s*")

valid_samples = [
    "salary,loan,deposit",
    "salary, loan, deposit",
    "account, savings, banking"
]

invalid_samples = [
    "",
    None,
    "salary",
    ",,,"
]

for text in valid_samples:
    tags = pattern.split(text)
    print("Tags:", tags)

for text in invalid_samples:
    if not isinstance(text, str):
        print("Invalid input:", text)
        continue

    if not text.strip():
        print("Empty tag list")
        continue

    tags = [tag.strip() for tag in pattern.split(text) if tag.strip()]

    if tags:
        print("Tags:", tags)
    else:
        print("Invalid tag list:", text)
```

## === Q10. Replace masked card digits for banking case 1; use sample reference ACC-1010 and explain the result. ===

```python
import re

# Sample reference: ACC-1010

pattern = re.compile(r"(?P<masked>\*{4,}\d{4})")

valid_samples = [
    "Card: ****1234",
    "Payment card: ******5678",
    "Used card ****9012 for payment"
]

invalid_samples = [
    "12345678",
    "****",
    "****123",
    "",
    None
]

def replace_masked_digits(text):
    if not isinstance(text, str):
        return text

    return pattern.sub(lambda match: "****" + match.group("masked")[-4:], text)


for text in valid_samples:
    print("Original:", text)
    print("Replaced:", replace_masked_digits(text))

for text in invalid_samples:
    if not isinstance(text, str):
        print("Invalid input:", text)
        continue

    result = replace_masked_digits(text)
    print("Result:", result)
```
