## === Q33. Process transfers while isolating bad rows for banking case 1; use sample reference ACC-1033 and explain the result. ===

```python
class InvalidTransactionError(Exception):
    """Raised when a transfer is invalid."""
    pass


account_number = "ACC-1033"
balance = 10000

transfers = [
    {"reference": "TXN-001", "amount": 2000},
    {"reference": "TXN-002", "amount": "abc"},
    {"reference": "TXN-003", "amount": 3000},
    {"reference": "TXN-004", "amount": -500}
]

successful_transfers = []
failed_transfers = []

for transfer in transfers:
    try:
        amount = float(transfer["amount"])

        if amount <= 0:
            raise InvalidTransactionError(
                "Transfer amount must be greater than zero."
            )

        if amount > balance:
            raise InvalidTransactionError(
                "Insufficient funds for this transfer."
            )

        # Update balance only after validation succeeds
        balance -= amount

        successful_transfers.append(transfer["reference"])

    except (ValueError, InvalidTransactionError) as error:
        failed_transfers.append(
            (transfer["reference"], str(error))
        )

print("Account Number:", account_number)
print("Final Balance:", balance)

print("\nSuccessful Transfers:")
for reference in successful_transfers:
    print(reference)

print("\nFailed Transfers:")
for reference, error in failed_transfers:
    print(reference, "-", error)
```

## === Q34. Retry a transient payment gateway for banking case 1; use sample reference ACC-1034 and explain the result. ===

```python
class TransientGatewayError(Exception):
    """Raised when the payment gateway temporarily fails."""
    pass


def payment_gateway():
    # Simulate a temporary gateway failure on the first two attempts
    payment_gateway.attempts += 1

    if payment_gateway.attempts < 3:
        raise TransientGatewayError("Payment gateway temporarily unavailable.")

    return True


payment_gateway.attempts = 0

account_number = "ACC-1034"
balance = 10000
payment_amount = 3000
max_retries = 3

for attempt in range(1, max_retries + 1):
    try:
        payment_gateway()

        # Update balance only after successful payment
        balance -= payment_amount

        print("Payment successful.")
        print("Account Number:", account_number)
        print("Payment Amount:", payment_amount)
        print("Remaining Balance:", balance)
        break

    except TransientGatewayError as error:
        print(f"Attempt {attempt} failed:", error)

        if attempt == max_retries:
            print("Payment failed after all retry attempts.")
```

## === Q40. Collect all kyc validation errors for banking case 1; use sample reference ACC-1040 and explain the result. ===

```python
class KYCValidationError(Exception):
    """Raised when KYC validation fails."""
    pass


def validate_kyc(customer):
    errors = []

    # Validate name
    if not customer.get("name", "").strip():
        errors.append("Name is required.")

    # Validate age
    age = customer.get("age")

    if age is None:
        errors.append("Age is required.")
    elif not isinstance(age, int):
        errors.append("Age must be an integer.")
    elif age < 18:
        errors.append("Customer must be at least 18 years old.")

    # Validate address
    if not customer.get("address", "").strip():
        errors.append("Address is required.")

    if errors:
        raise KYCValidationError(errors)

    return True


account_number = "ACC-1040"

customer = {
    "name": "",
    "age": 16,
    "address": ""
}

try:
    validate_kyc(customer)

    print("KYC validation successful.")
    print("Account Number:", account_number)

except KYCValidationError as error:
    print("KYC validation failed.")

    for message in error.args[0]:
        print("-", message)
```

## === Q44. Retry a transient payment gateway for banking case 2; use sample reference ACC-1044 and explain the result. ===

```python
class TransientGatewayError(Exception):
    """Raised when the payment gateway temporarily fails."""
    pass


def payment_gateway():
    # Simulate a temporary failure on the first two attempts
    payment_gateway.attempts += 1

    if payment_gateway.attempts < 3:
        raise TransientGatewayError(
            "Payment gateway temporarily unavailable."
        )

    return True


payment_gateway.attempts = 0

account_number = "ACC-1044"
balance = 10000
payment_amount = 3000
max_retries = 3

for attempt in range(1, max_retries + 1):
    try:
        # Try the payment gateway
        payment_gateway()

        # Update balance only after successful payment
        balance -= payment_amount

        print("Payment successful.")
        print("Account Number:", account_number)
        print("Payment Amount:", payment_amount)
        print("Remaining Balance:", balance)

        break

    except TransientGatewayError as error:
        print(f"Attempt {attempt} failed:", error)

        if attempt == max_retries:
            print("Payment failed after all retry attempts.")
```

## === Q46. Return a structured failure result for banking case 2; use sample reference ACC-1046 and explain the result. ===

```python
class InvalidTransactionError(Exception):
    """Raised when a transaction is invalid."""
    pass


def process_transaction(account_number, balance, amount):
    try:
        # Validate transaction amount
        if amount <= 0:
            raise InvalidTransactionError(
                "Transaction amount must be greater than zero."
            )

        # Check available balance
        if amount > balance:
            raise InvalidTransactionError(
                "Insufficient funds."
            )

        # Update balance only after validation succeeds
        balance -= amount

        return {
            "status": "SUCCESS",
            "account_number": account_number,
            "amount": amount,
            "remaining_balance": balance
        }

    except InvalidTransactionError as error:
        return {
            "status": "FAILED",
            "account_number": account_number,
            "amount": amount,
            "error": str(error),
            "remaining_balance": balance
        }


account_number = "ACC-1046"
balance = 5000
withdrawal_amount = 7000

result = process_transaction(
    account_number,
    balance,
    withdrawal_amount
)

print(result)
```

## === Q49. Separate validation and recovery for banking case 2; use sample reference ACC-1049 and explain the result. ===

```python
class InvalidTransactionError(Exception):
    """Raised when a transaction fails validation."""
    pass


def validate_transaction(amount):
    """Validate the transaction amount."""
    if amount <= 0:
        raise InvalidTransactionError(
            "Transaction amount must be greater than zero."
        )

    return True


def process_transaction(account_number, balance, amount):
    """Process the transaction and handle validation failures."""
    try:
        # Validation is handled separately
        validate_transaction(amount)

    except InvalidTransactionError as error:
        # Recovery: return a structured failure result
        return {
            "status": "FAILED",
            "account_number": account_number,
            "error": str(error),
            "balance": balance
        }

    # Update balance only after validation succeeds
    balance -= amount

    return {
        "status": "SUCCESS",
        "account_number": account_number,
        "amount": amount,
        "balance": balance
    }


account_number = "ACC-1049"
balance = 10000
withdrawal_amount = -500

result = process_transaction(
    account_number,
    balance,
    withdrawal_amount
)

print(result)
```

## === Q50. Collect all kyc validation errors for banking case 2; use sample reference ACC-1050 and explain the result. ===

```python
class KYCValidationError(Exception):
    """Raised when KYC validation fails."""
    pass


def validate_kyc(customer):
    errors = []

    # Validate name
    if not customer.get("name", "").strip():
        errors.append("Name is required.")

    # Validate age
    age = customer.get("age")

    if age is None:
        errors.append("Age is required.")
    elif not isinstance(age, int):
        errors.append("Age must be an integer.")
    elif age < 18:
        errors.append("Customer must be at least 18 years old.")

    # Validate address
    if not customer.get("address", "").strip():
        errors.append("Address is required.")

    # Raise one exception containing all validation errors
    if errors:
        raise KYCValidationError(errors)

    return True


account_number = "ACC-1050"

customer = {
    "name": "",
    "age": 16,
    "address": ""
}

try:
    validate_kyc(customer)

    print("KYC validation successful.")
    print("Account Number:", account_number)

except KYCValidationError as error:
    print("KYC validation failed.")

    for message in error.args[0]:
        print("-", message)
```
